```python
import os
import subprocess


BASELINE_MODELS = [
    "GigaChat-3-Ultra"
]

GIGACHAT_TOKEN = "YOUR_GIGACHAT_TOKEN" # Ваш API ключ для GigaChat
BASE_URL = "https://gigachat.ift.sberdevices.ru/v1"

ENV_PATH = "/home/jovyan/.mlspace/envs/conda_env" # Путь до папки с окружением для запуска MERA. Если это конда-окружение с именем conda_env, то указываете путь до этой папки.
HF_HOME = "/home/jovyan/.cache/huggingface" # Если необходимо поменять, например на локальный кеш
WORKING_DIR = "" # Абсолютный путь до репозитория MERA
OUTPUT_PATH = "" # Путь до папки для сохранения результатов. Может быть любой
CACHE_DIR = "" # Путь до папки для сохранения запросов. Может быть любой
HF_DATASETS_CACHE= "" # Путь до папки для сохранения кеша датасетов. Может быть любой


task = "simplear" # Имя вашей задачи
num_fewshot = 0 # Число фью-шотов для задачи. По умолчанию 0
max_gen_toks = 16384 # Максимальное количество токенов для генерации ответа

for model_name in BASELINE_MODELS:
    model_author, short_name = model_name.split('/')
    path = os.path.join(OUTPUT_PATH, model_author, short_name, f"{model_name.replace('/', '__')}")

    exp_name = f"#eval_mera_{model_name}_{task}"

    script_name = os.path.join(WORKING_DIR, "scripts", "gigachat_api.sh")

    command = " ".join([
        f"GIGACHAT_TOKEN={GIGACHAT_TOKEN}",
        f"BASE_URL={BASE_URL}",
        f"ENV_PATH={ENV_PATH}",
        f"HF_HOME={HF_HOME}",
        f"WORKING_DIR={WORKING_DIR}",
        f"CACHE_DIR={CACHE_DIR}",
        f"HF_DATASETS_CACHE={HF_DATASETS_CACHE}",
        f"OUTPUT_PATH={OUTPUT_PATH}",
        f"MODEL_NAME={model_name}",
        f"TASK_NAME={task}",
        f"NUM_FEWSHOT={num_fewshot}",
        f"MAX_GEN_TOKS={max_gen_toks}",
        f"bash {script_name}" 
    ])
    os.makedirs(f"{OUTPUT_PATH}/logs", exist_ok=True)

    process = subprocess.run(command, shell=True, executable="/usr/bin/bash", stdout=open(f"{OUTPUT_PATH}/logs/{short_name}_{task}.log", "w"))
    
```
